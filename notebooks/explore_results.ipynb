# Explore a saved study

Loads `results/qwen2.5-7b/judgments.jsonl` after `make study`. The teaching narrative lives in `walkthrough_local_judge.ipynb`.


```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / "pyproject.toml").exists():
    ROOT = Path("..").resolve()
sys.path.insert(0, str(ROOT / "src"))

from preference_consistency.plots import (
    figure_condition_rates,
    figure_verbosity,
    highlights,
    load_judgments,
)
from IPython.display import display

path = ROOT / "results/qwen2.5-7b/judgments.jsonl"
df = load_judgments(path)
df.groupby("condition").size()
```

```python
valid = df.loc[~df["parse_error"]]
print(valid.groupby("condition")["correct"].mean())
print("P(first)", valid.groupby("condition")["picked_first"].mean())
for name, item in highlights(df):
    print(f"{name}: {item.point:.3f} ({item.k}/{item.n})")
```

```python
display(figure_condition_rates(df, "accuracy", "Accuracy vs HH label"))
display(figure_condition_rates(df, "first", "P(verdict = A)"))
fig = figure_verbosity(df, "Verbosity bloat")
if fig is not None:
    display(fig)
```
