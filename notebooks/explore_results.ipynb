# Explore preference-consistency results

Reads `results/judgments.jsonl` after `make smoke` or a real Ollama run.

```python
from pathlib import Path
import pandas as pd

path = Path("results/judgments.jsonl")
df = pd.read_json(path, lines=True)
df.head()
```

```python
valid = df.loc[~df["parse_error"]]
print(valid.groupby("condition")["correct"].mean())
print("P(first)", valid.groupby("condition")["picked_first"].mean())
```
